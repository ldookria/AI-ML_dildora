```python
plt.figure(figsize=(8, 5))

plt.hist(df["age"], bins=15, edgecolor="black")

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Records")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="income",
    order=["<=50K", ">50K"]
)

plt.title("Number of People in Each Income Group")
plt.xlabel("Income Group")
plt.ylabel("Number of Records")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="income",
    y="hours.per.week",
    order=["<=50K", ">50K"]
)

plt.title("Hours Worked per Week by Income Group")
plt.xlabel("Income Group")
plt.ylabel("Hours per Week")

plt.show()
```

```python
occupation_counts = (
    df["occupation"]
    .value_counts()
    .head(8)
    .sort_values()
)

plt.figure(figsize=(9, 5))

occupation_counts.plot.barh()

plt.title("Eight Most Common Occupations")
plt.xlabel("Number of Records")
plt.ylabel("Occupation")

plt.show()
```
